```python
# Import pandas for data analysis
import pandas as pd

# Load the national panel dataset
# ../ moves one folder up from notebooks/ to the project root
df = pd.read_csv("../processed/national_all_years_2011units.csv")

# --------------------------------------------------
# 1. Check urban and non-urban unit distribution
# --------------------------------------------------
print("\n========== URBAN UNIT DISTRIBUTION ==========")

print(
    df.groupby("year")["is_urban_unit"].value_counts()
)

# --------------------------------------------------
# 2. Display unique urban geographical units
# --------------------------------------------------
print("\n========== URBAN UNITS ==========")

urban = df[df["is_urban_unit"] == True]

print(
    urban[["unit_id", "unit_name", "state"]]
    .drop_duplicates()
    .to_string(index=False)
)

# --------------------------------------------------
# 3. Display all records with known anomalies
# --------------------------------------------------
print("\n========== ANOMALOUS RECORDS ==========")

anomalies = df[df["known_anomaly"] == True]

print(
    anomalies[
        ["state", "unit_name", "year", "anomaly_reason"]
    ].to_string(index=False)
)

# --------------------------------------------------
# 4. Check duplicate geographical unit-year records
# --------------------------------------------------
print("\n========== DUPLICATE CHECK ==========")

duplicates = df.duplicated(["unit_id", "year"]).sum()

print("Duplicate records:", duplicates)

# --------------------------------------------------
# 5. Check missing population values
# --------------------------------------------------
print("\n========== MISSING POPULATION ==========")

print("Missing population:", df["population"].isna().sum())

# --------------------------------------------------
# 6. Check records per geographical unit
# --------------------------------------------------
print("\n========== RECORDS PER UNIT ==========")

print(
    df.groupby("unit_id").size().value_counts().sort_index()
)

# --------------------------------------------------
# 7. Check year-wise record distribution
# --------------------------------------------------
print("\n========== YEAR DISTRIBUTION ==========")

print(df["year"].value_counts().sort_index())

# --------------------------------------------------
# 8. Final dataset dimensions
# --------------------------------------------------
print("\n========== DATASET SUMMARY ==========")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Unique geographical units:", df["unit_id"].nunique())

print("\nNational dataset inspection completed!")
```
